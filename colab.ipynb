## 2 · Get the code
Either put your GitHub repo address below (after you have pushed it), **or leave it empty** and a file picker will ask you to upload `video-captioner.zip`.

```
REPO_URL = ""   # e.g. "https://github.com/YOUR-USERNAME/video-captioner.git"

import os, glob, subprocess, zipfile
os.chdir("/content")
if REPO_URL:
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, "video-captioner"], check=True)
else:
    from google.colab import files
    up = files.upload()                      # choose video-captioner.zip
    for name in up:
        zipfile.ZipFile(name).extractall("/content/project")
hits = glob.glob("/content/**/app.py", recursive=True)
hits = [h for h in hits if "site-packages" not in h]
assert hits, "app.py not found - did the upload/clone work?"
os.chdir(os.path.dirname(hits[0]))
print("Working in:", os.getcwd())
```

## 4 · Start the app
When it prints a link like `https://xxxx.gradio.live`, click it. It opens the app in a new tab.
Keep this cell running while you use the app. In the app choose accuracy **large-v3** for the best Telugu (fast on the GPU).

```
import os, glob, site
os.environ["GRADIO_SHARE"] = "1"
# let the speech model find the GPU libraries installed above
libs = [d for sp in site.getsitepackages() for d in glob.glob(sp + "/nvidia/*/lib")]
os.environ["LD_LIBRARY_PATH"] = ":".join(libs + [os.environ.get("LD_LIBRARY_PATH", "")])
!python app.py
```
